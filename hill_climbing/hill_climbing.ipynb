```python
class Problem:
    def __init__(self, initial, goal):
        self.initial = initial 
        self.goal = goal 
    
    def actions(self, state):
        raise NotImplementedError

    def result(self, state, action):
        raise NotImplementedError

    def is_goal(self, state):
        return self.goal == state
    
    def action_cost(self, state1, action, state2):
        return 1
   
    def h(self, state):
        return 0
```

```python
class GraphProblem(Problem):
    def __init__(self, initial, goal, graph, heuristics=None):
        super().__init__(initial, goal)
        self.graph = graph
        self.heuristics = heuristics if heuristics else {}
    
    def actions(self, state):
        return list(self.graph[state].keys())

    def result(self, state, action):
        return action

    def action_cost(self, state1, action, state2):
        return self.graph[state1][state2]
        
    def h(self, state):
        return self.heuristics.get(state, float('inf'))
```

```python
def hill_climbing(problem):
    """
    Algoritmo de ascenso de colinas (Hill-Climbing).
    Como buscamos minimizar la distancia a la meta (costo/heurística), 
    elegimos el vecino con el MENOR valor de h(s).
    """
    current = problem.initial
    path = [current] 
    
    while True:
        neighbors = problem.actions(current)
        if not neighbors:
            break
            
        next_state = min(neighbors, key=lambda state: problem.h(problem.result(current, state)))
        
        if problem.h(next_state) >= problem.h(current):
            break
            
        current = next_state
        path.append(current)
        
        if problem.is_goal(current):
            break
            
    return path, current
```
